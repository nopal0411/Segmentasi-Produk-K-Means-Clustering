## ðŸ§¹ Tahap 3: Data Preprocessing (KDD Stage 2)

Pembersihan data yang mencakup:
- Deteksi & penanganan missing values
- Penghapusan duplikat
- Normalisasi nama kolom
- Konversi tipe data numerik


```python
# â”€â”€ Salin dataframe agar tidak merusak data asli â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
df = df_raw.copy()

# Normalisasi nama kolom: lowercase, hapus spasi
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
    .str.replace('/', '_')
    .str.replace('-', '_')
)

print("ðŸ“Œ Kolom setelah normalisasi:")
print(df.columns.tolist())
```

```python
# â”€â”€ Cek Missing Values â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing': missing, 'Persentase (%)': missing_pct})
missing_df = missing_df[missing_df['Missing'] > 0]

if missing_df.empty:
    print("âœ… Tidak ada missing values!")
else:
    print("âš ï¸  Missing Values yang Ditemukan:")
    print(missing_df)
```

```python
# â”€â”€ Cek Duplikat â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
n_dup = df.duplicated().sum()
print(f"ðŸ” Duplikat ditemukan: {n_dup} baris")
if n_dup > 0:
    df.drop_duplicates(inplace=True)
    df.reset_index(drop=True, inplace=True)
    print(f"âœ… Duplikat dihapus. Sisa data: {len(df)} baris")
```

```python
# â”€â”€ Identifikasi Kolom Numerik Otomatis â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Cari kolom yang kemungkinan berisi: sell_in, sell_out, sales, stok, penjualan
import re

def find_col(df_cols, keywords):
    """Temukan kolom pertama yang mengandung salah satu keyword."""
    for kw in keywords:
        for col in df_cols:
            if kw in col:
                return col
    return None

cols = df.columns.tolist()

col_nama     = find_col(cols, ['nama', 'produk', 'product', 'item'])
col_sales    = find_col(cols, ['sales', 'revenue', 'nilai', 'omzet'])
col_sell_in  = find_col(cols, ['sell_in', 'sellin', 'masuk', 'in'])
col_sell_out = find_col(cols, ['sell_out', 'sellout', 'keluar', 'out'])
col_stok     = find_col(cols, ['stok', 'stock', 'sisa', 'qty', 'quantity'])
col_jual     = find_col(cols, ['jual', 'terjual', 'unit', 'penjualan'])
col_expired  = find_col(cols, ['expired', 'exp', 'kadaluarsa'])

print("ðŸ” Mapping Kolom Terdeteksi:")
print(f"  Nama Produk : {col_nama}")
print(f"  Sales       : {col_sales}")
print(f"  Sell In     : {col_sell_in}")
print(f"  Sell Out    : {col_sell_out}")
print(f"  Stok        : {col_stok}")
print(f"  Penjualan   : {col_jual}")
print(f"  Expired     : {col_expired}")
```

```python
# â”€â”€ Konversi Kolom Numerik â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
num_candidates = [col_sales, col_sell_in, col_sell_out, col_stok, col_jual]
num_cols = [c for c in num_candidates if c is not None]

for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors='coerce')

# Isi missing values numerik dengan median
for c in num_cols:
    if df[c].isnull().sum() > 0:
        median_val = df[c].median()
        df[c].fillna(median_val, inplace=True)
        print(f"  â†ª {c}: {df[c].isnull().sum()} NaN diisi dengan median ({median_val:.2f})")

print(f"\nâœ… Preprocessing selesai. Shape akhir: {df.shape}")
df.head()
```

## ðŸ”§ Tahap 4: Data Transformation & Feature Engineering (KDD Stage 3)

Feature engineering yang dilakukan:
- **Rasio Sell Out / Sell In** â†’ Mengukur efisiensi distribusi  
- **Tingkat perputaran stok** â†’ Seberapa cepat produk habis  
- **Normalisasi** menggunakan MinMaxScaler


```python
# â”€â”€ Feature Engineering â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Rasio sell out / sell in (efisiensi distribusi)
if col_sell_in and col_sell_out:
    df['rasio_sell_out_in'] = np.where(
        df[col_sell_in] > 0,
        df[col_sell_out] / df[col_sell_in],
        0
    )
    print("âœ… Fitur 'rasio_sell_out_in' dibuat")

# Tingkat perputaran stok
if col_jual and col_stok:
    df['perputaran_stok'] = np.where(
        (df[col_stok] + df[col_jual]) > 0,
        df[col_jual] / (df[col_stok] + df[col_jual]),
        0
    )
    print("âœ… Fitur 'perputaran_stok' dibuat")

# Fitur dari expired
if col_expired:
    try:
        df[col_expired] = pd.to_datetime(df[col_expired], errors='coerce')
        today = pd.Timestamp('2025-12-31')
        df['sisa_hari_expired'] = (df[col_expired] - today).dt.days
        df['sisa_hari_expired'] = df['sisa_hari_expired'].fillna(df['sisa_hari_expired'].median())
        df['sisa_hari_expired'] = df['sisa_hari_expired'].clip(lower=0)
        print("âœ… Fitur 'sisa_hari_expired' dibuat")
    except Exception as e:
        print(f"âš ï¸  Gagal membuat fitur expired: {e}")

df.head()
```

```python
# â”€â”€ Pilih fitur untuk clustering â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
feature_candidates = [col_sales, col_sell_in, col_sell_out, col_stok, col_jual,
                      'rasio_sell_out_in', 'perputaran_stok', 'sisa_hari_expired']

features = [f for f in feature_candidates if f is not None and f in df.columns]
print("ðŸŽ¯ Fitur yang digunakan:")
for f in features:
    print(f"  - {f}")

X = df[features].copy()
print(f"\nðŸ“ Shape matriks fitur: {X.shape}")
```

```python
# â”€â”€ Visualisasi Distribusi Sebelum Normalisasi â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
n_feat = len(features)
fig, axes = plt.subplots(1 if n_feat <= 3 else 2, 
                          n_feat if n_feat <= 3 else (n_feat+1)//2, 
                          figsize=(15, 8))
axes = np.array(axes).flatten()

for i, feat in enumerate(features):
    axes[i].hist(X[feat].dropna(), bins=20, color='steelblue', edgecolor='white', alpha=0.8)
    axes[i].set_title(f'Distribusi: {feat}', fontsize=11)
    axes[i].set_xlabel('Nilai')
    axes[i].set_ylabel('Frekuensi')

# Sembunyikan axes kosong
for j in range(i+1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('ðŸ“Š Distribusi Fitur SEBELUM Normalisasi', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('distribusi_sebelum_normalisasi.png', dpi=150, bbox_inches='tight')
plt.show()
print("ðŸ’¾ Tersimpan: distribusi_sebelum_normalisasi.png")
```

```python
# â”€â”€ Normalisasi MinMaxScaler â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=features)

# Visualisasi distribusi setelah normalisasi
fig, axes = plt.subplots(1 if n_feat <= 3 else 2,
                          n_feat if n_feat <= 3 else (n_feat+1)//2,
                          figsize=(15, 8))
axes = np.array(axes).flatten()

for i, feat in enumerate(features):
    axes[i].hist(X_scaled_df[feat], bins=20, color='darkorange', edgecolor='white', alpha=0.8)
    axes[i].set_title(f'Distribusi: {feat}', fontsize=11)
    axes[i].set_xlabel('Nilai (0â€“1)')
    axes[i].set_ylabel('Frekuensi')

for j in range(i+1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('ðŸ“Š Distribusi Fitur SESUDAH Normalisasi (MinMaxScaler)', fontsize=14, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig('distribusi_sesudah_normalisasi.png', dpi=150, bbox_inches='tight')
plt.show()
print("ðŸ’¾ Tersimpan: distribusi_sesudah_normalisasi.png")
```

## ðŸ“ Tahap 5: Menentukan Jumlah Cluster Optimal (KDD Stage 4)

Dua metode digunakan:
- **Elbow Method** â†’ Inertia (WCSS) vs jumlah cluster
- **Silhouette Score** â†’ Kualitas pemisahan cluster


```python
# â”€â”€ Elbow Method + Silhouette Score â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
K_range = range(2, 11)
inertias = []
silhouettes = []

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X_scaled, labels))
    print(f"  k={k:2d}  |  Inertia: {km.inertia_:12.2f}  |  Silhouette: {silhouette_score(X_scaled, labels):.4f}")
```

```python
# â”€â”€ Plot Elbow + Silhouette â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Elbow
ax1.plot(list(K_range), inertias, 'bo-', linewidth=2, markersize=7)
ax1.axvline(x=3, color='red', linestyle='--', alpha=0.7, label='k=3 (optimal)')
ax1.set_title('ðŸ”· Elbow Method', fontsize=13, fontweight='bold')
ax1.set_xlabel('Jumlah Cluster (k)')
ax1.set_ylabel('Inertia (WCSS)')
ax1.legend()
ax1.grid(True, alpha=0.4)

# Silhouette
ax2.plot(list(K_range), silhouettes, 'rs-', linewidth=2, markersize=7)
best_k = list(K_range)[np.argmax(silhouettes)]
ax2.axvline(x=best_k, color='blue', linestyle='--', alpha=0.7, label=f'k={best_k} (terbaik)')
ax2.set_title('ðŸ”¶ Silhouette Score', fontsize=13, fontweight='bold')
ax2.set_xlabel('Jumlah Cluster (k)')
ax2.set_ylabel('Silhouette Score')
ax2.legend()
ax2.grid(True, alpha=0.4)

plt.suptitle('ðŸ“ Penentuan Jumlah Cluster Optimal', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('elbow_silhouette.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"\nðŸ“Œ Silhouette terbaik: k = {best_k}  ({max(silhouettes):.4f})")
print(f"ðŸ“Œ Untuk interpretasi bisnis, kita gunakan k = 3 (Fast/Medium/Slow Moving)")
```

## ðŸ¤– Tahap 6: K-Means Clustering (KDD Stage 4 â€“ Data Mining)

Melatih model K-Means dengan k=3 untuk menghasilkan 3 segmen produk.


```python
# â”€â”€ Train K-Means k=3 â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
K_OPTIMAL = 3
km_final = KMeans(n_clusters=K_OPTIMAL, random_state=42, n_init=10, max_iter=300)
df['cluster'] = km_final.fit_predict(X_scaled)

sil = silhouette_score(X_scaled, df['cluster'])
print(f"âœ… K-Means selesai!")
print(f"   Jumlah cluster : {K_OPTIMAL}")
print(f"   Inertia        : {km_final.inertia_:,.2f}")
print(f"   Silhouette Score: {sil:.4f}")
print(f"\nðŸ“Š Distribusi cluster:")
print(df['cluster'].value_counts().sort_index())
```

## ðŸ·ï¸ Tahap 7: Interpretasi Cluster â†’ Fast / Medium / Slow Moving (KDD Stage 5)

Setiap cluster dianalisis berdasarkan rata-rata fitur untuk menentukan label kategori:
- **Fast Moving**: Volume penjualan & sell out tinggi
- **Medium Moving**: Volume penjualan & sell out sedang  
- **Slow Moving**: Volume penjualan & sell out rendah


```python
# â”€â”€ Analisis centroid per cluster â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
cluster_summary = df.groupby('cluster')[features].mean()
print("ðŸ“Š Rata-rata fitur per cluster:")
cluster_summary.round(2)
```

```python
# â”€â”€ Pemetaan cluster ke kategori bisnis â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# Pilih kolom utama untuk ranking (prioritas: sell_out > penjualan > sales)
rank_col = col_sell_out or col_jual or col_sales

if rank_col:
    cluster_rank = df.groupby('cluster')[rank_col].mean().sort_values(ascending=False)
    print(f"ðŸ“Œ Ranking cluster berdasarkan '{rank_col}':")
    print(cluster_rank.round(2))
    
    ranking = cluster_rank.index.tolist()  # [cluster_tertinggi, sedang, terendah]
    mapping = {
        ranking[0]: 'Fast Moving',
        ranking[1]: 'Medium Moving',
        ranking[2]: 'Slow Moving',
    }
else:
    # Fallback: ranking berdasarkan rata-rata semua fitur
    cluster_mean = X_scaled_df.copy()
    cluster_mean['cluster'] = df['cluster'].values
    cluster_ranks = cluster_mean.groupby('cluster').mean().mean(axis=1).sort_values(ascending=False)
    ranking = cluster_ranks.index.tolist()
    mapping = {ranking[0]: 'Fast Moving', ranking[1]: 'Medium Moving', ranking[2]: 'Slow Moving'}

df['kategori'] = df['cluster'].map(mapping)

print("\nðŸ·ï¸  Mapping Cluster â†’ Kategori:")
for k, v in mapping.items():
    print(f"  Cluster {k} â†’ {v}")

print("\nðŸ“Š Distribusi Kategori:")
print(df['kategori'].value_counts())
```

## ðŸ“Š Tahap 8: Visualisasi Hasil Clustering (KDD Stage 5)

## ðŸ’¡ Tahap 9: Business Insight & Rekomendasi (KDD Stage 6 â€“ Knowledge)


```python
# â”€â”€ Ringkasan statistik per kategori â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("=" * 70)
print("ðŸ“Š RINGKASAN SEGMENTASI PRODUK")
print("=" * 70)

for kat in ['Fast Moving', 'Medium Moving', 'Slow Moving']:
    subset = df[df['kategori'] == kat]
    print(f"\n{'â”'*60}")
    print(f"  ðŸ·ï¸  {kat.upper()}  ({len(subset)} produk)")
    print(f"{'â”'*60}")
    for feat in features:
        print(f"    {feat:<30}: {subset[feat].mean():>12.2f}  (avg)")
    
    if col_nama and col_nama in df.columns:
        top3 = subset.nlargest(3, rank_col or features[0])[col_nama].values if rank_col else []
        if len(top3) > 0:
            print(f"\n  ðŸ” Top produk: {', '.join(str(p) for p in top3)}")
```

```python
# â”€â”€ Insight Bisnis Terstruktur â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\n" + "="*70)
print("ðŸ’¡ BUSINESS INSIGHT & REKOMENDASI")
print("="*70)

insights = {
    "ðŸ”´ Fast Moving": [
        "Produk dengan perputaran stok tercepat dan demand tinggi.",
        "â†’ Prioritaskan restock & ketersediaan produk.",
        "â†’ Pastikan rantai distribusi tidak putus.",
        "â†’ Tingkatkan kapasitas produksi sesuai demand.",
    ],
    "ðŸŸ¡ Medium Moving": [
        "Produk dengan perputaran stok sedang, potensi besar.",
        "â†’ Lakukan promosi dan bundling untuk meningkatkan demand.",
        "â†’ Monitor tren penjualan secara berkala.",
        "â†’ Evaluasi harga dibanding kompetitor.",
    ],
    "ðŸŸ¢ Slow Moving": [
        "Produk dengan perputaran stok lambat, berisiko overstock.",
        "â†’ Terapkan diskon atau flash sale untuk menghabiskan stok.",
        "â†’ Evaluasi kelayakan produk di pasar.",
        "â†’ Pertimbangkan strategi bundling dengan Fast Moving.",
        "â†’ Perhatikan tanggal expired untuk mencegah kerugian.",
    ],
}

for kategori, poin in insights.items():
    print(f"\n{kategori}")
    for p in poin:
        print(f"  {p}")

print("\n" + "="*70)
```

```python
# â”€â”€ Pertanyaan Analitis â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
print("\nðŸ“ˆ JAWABAN PERTANYAAN ANALITIS")
print("="*70)

# 1. Produk perputaran stok tercepat
if rank_col and col_nama and col_nama in df.columns:
    top_fast = df[df['kategori'] == 'Fast Moving'].nlargest(5, rank_col)[[col_nama, rank_col, 'kategori']]
    print(f"\n1. Produk perputaran stok TERCEPAT (Top 5 berdasarkan {rank_col}):")
    print(top_fast.to_string(index=False))

# 2. Produk potensi penumpukan stok
if col_stok and col_stok in df.columns:
    slow = df[df['kategori'] == 'Slow Moving'].nlargest(5, col_stok)
    cols_show = [c for c in [col_nama, col_stok, 'kategori'] if c in df.columns]
    print(f"\n2. Produk berpotensi OVERSTOCK (Top 5 Slow Moving, stok terbesar):")
    print(slow[cols_show].to_string(index=False))

# 3. Korelasi sell in vs sell out
if col_sell_in and col_sell_out:
    corr_val = df[[col_sell_in, col_sell_out]].corr().iloc[0, 1]
    print(f"\n3. Korelasi Sell In â†” Sell Out: {corr_val:.4f}")
    if corr_val > 0.8:
        print("   â†’ Korelasi sangat kuat: distribusi berjalan efisien")
    elif corr_val > 0.5:
        print("   â†’ Korelasi sedang: ada ketidakseimbangan distribusi")
    else:
        print("   â†’ Korelasi lemah: perlu evaluasi strategi distribusi")

# 4. Cluster paling optimal untuk profit
if col_sales:
    profit_by_cat = df.groupby('kategori')[col_sales].sum().sort_values(ascending=False)
    print(f"\n4. Kontribusi Sales per Kategori (Rp):")
    for kat, val in profit_by_cat.items():
        print(f"   {kat:<18}: Rp {val:>15,.0f}")
    best = profit_by_cat.index[0]
    print(f"   â†’ Kategori paling optimal: {best}")
```
